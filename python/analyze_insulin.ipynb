Try using regex to programmatically strip the file of ORIGIN, its numbers, the two slashes (//), spaces, and line breaks or return carriages. You could also confirm programmatically that the file has 110 characters.

### Using REGEX

```python
import re
```

```python
file = open(r'preproinsulin-seq.txt').read()
print(file)
```

```python
string_list = re.split(r"\s+", file)
print(string_list)
```

#### Step 1: Extract the sequence region

```python
# Extract the sequence between 'ORIGIN' and '//'
# . means any character.
# * means zero or more times.
# ? makes it non-greedy
# () captures the group
extracted = re.search(r'ORIGIN\s+(.*?)\s*//', file, re.DOTALL).group(1) 
print(extracted)
```

#### Step 2: Strip the unwanted characters using regex substitution

```python
# Remove all digits and whitespace
# \d matches any digit
# \s matches any whitespace character
# '' is the replacement string (empty string)
cleaned = re.sub(r'[\d\s]', '', extracted)
print(cleaned)
```

#### Step 3: Verify the character count

```python
char_count = len(cleaned)
print(f"Character count: {char_count}")
```

#### Result

```python
print(f"Cleaned sequence: {cleaned}")
print(f"Character count: {char_count}")
print(f"Is equal to 110: {char_count == 110}")
```
